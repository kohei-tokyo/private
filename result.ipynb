```python
from resultGAN import predict
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
images = []

encoder_names = ["efficientnet-b4", "resnet50"]
decoder_attention_types = [None, "scse"]
for encoder_name in encoder_names:
    for decoder_attention_type in decoder_attention_types:
        images.append(predict(
            r"C:\data\ER_crop_256\3",
            name=f"{encoder_name}_encoder_{str(decoder_attention_type)}_decorder_ER_1027",
            encoder_name=encoder_name,
            decoder_attention_type=decoder_attention_type,
            test_id="lpips",
            img_size=256,
            n_result=4,
            device="cuda" if torch.cuda.is_available() else "cpu"
        ))
```

```python
target = 3
n = 4
images_n = 4
for i in range(n):
    fig, axs = plt.subplots(2, 3, figsize=(15, 9))
    axs[0,0].imshow(images[0][2][i][0], cmap="viridis", vmin=0, vmax=1)
    axs[0,0].axis('off')
    axs[0,0].set_title('target')
    axs[1,0].imshow(images[0][2][i][0], cmap="viridis", vmin=0, vmax=1)
    axs[1,0].axis('off')
    axs[1,0].set_title('target')
    for j in range(images_n):
        axs[j//2, j%2+1].imshow(images[j][3][i][0], cmap="viridis", vmin=0, vmax=1)
        axs[j//2, j%2+1].axis('off')
        axs[j//2, j%2+1].set_title(f"{encoder_names[j//2]}_{decoder_attention_types[j%2]}")
    plt.tight_layout()
    plt.savefig(f"D:\\result_images\\ER_encoder_decoder_{i}", dpi=300)
    plt.show()
```
